# Notebook 9 – Grouping & Aggregation

# 1. groupby()

## Step 1 – Understand the Concept

`groupby()` is used to divide data into groups based on one or more columns.

After creating groups, we can perform operations such as sum, mean, count, minimum, and maximum on each group.

### Why is it needed?

It helps us compare and summarize data category-wise instead of analyzing the entire dataset as one group.

## Step 2 – Demonstrate the Concept

### Simple Example

We can group students based on their city and calculate the average marks for each city.

### Real-World Example

A company can group customers based on their city and study their purchase behavior.

### Business Example

A business can group sales by product category to find the total sales for each category.

### AI/ML Use Case

Grouping can help create aggregated features such as average purchase amount per customer.

### Data Analytics Use Case

Analysts use grouping to compare performance across categories, regions, products, or departments.


## Step 3: Implement the Concept

```python
import pandas as pd

df = pd.DataFrame({
    "Name": ["Arun", "Priya", "Kumar", "Anitha", "Rahul", "Divya"],
    "City": ["Chennai", "Mumbai", "Chennai", "Delhi", "Mumbai", "Chennai"],
    "Product": ["Laptop", "Phone", "Phone", "Laptop", "Tablet", "Laptop"],
    "Sales": [50000, 30000, 25000, 45000, 20000, 55000]
})

city_sales = df.groupby("City")["Sales"].sum()

print(city_sales)
```

## Code Explanation

- `groupby("City")` divides the data into groups based on City.
- `["Sales"]` selects the Sales column.
- `sum()` calculates the total sales for each city.
- The result shows the total sales separately for Chennai, Mumbai, and Delhi.

### AI/ML Use

Grouped values can be used to create aggregated features for machine learning models.

### Data Analytics Use

It helps compare business performance across different groups.

# 2. agg()

## Step 1 – Understand the Concept

`agg()` is used to apply one or more aggregation functions to grouped data.

Common aggregation functions include:

- `sum()`
- `mean()`
- `min()`
- `max()`
- `count()`

### Why is it needed?

`agg()` allows us to calculate multiple statistics at the same time.

## Step 2 – Demonstrate the Concept

### Simple Example

For each city, we can calculate total sales, average sales, and maximum sales.

### Real-World Example

A company can calculate multiple statistics for customers in different regions.

### Business Example

A sales manager can view total, average, minimum, and maximum sales for each product.

### AI/ML Use Case

Multiple aggregated statistics can be used to create meaningful features for machine learning.

### Data Analytics Use Case

Analysts can generate a detailed summary of each group using multiple aggregation functions.


## Step 3: Implement the Concept

```python
city_summary = df.groupby("City")["Sales"].agg(
    ["sum", "mean", "min", "max", "count"]
)

print(city_summary)
```

## Code Explanation

- `groupby("City")` creates groups based on City.
- `agg()` applies multiple functions to the Sales column.
- `sum` calculates total sales.
- `mean` calculates average sales.
- `min` and `max` find the lowest and highest sales.
- `count` counts the number of records in each city.

### AI/ML Use

These aggregated values can become useful features for predictive models.

### Data Analytics Use

It provides multiple summary statistics for each category.

# 3. transform()

## Step 1 – Understand the Concept

`transform()` performs an operation on each group but returns a result with the same number of rows as the original DataFrame.

Unlike `groupby().sum()`, which produces one result per group, `transform()` broadcasts the group-level result back to the original rows.

### Why is it needed?

It is useful when we want to compare each individual record with a group-level value such as the group average.

## Step 2 – Demonstrate the Concept

### Simple Example

We can calculate the average sales for each city and add that average to every row belonging to that city.

### Real-World Example

A company can compare each customer's purchase amount with the average purchase amount of their city.

### Business Example

A sales employee's sales can be compared with the average sales of their region.

### AI/ML Use Case

`transform()` can create group-based features while keeping the original number of rows.

### Data Analytics Use Case

Analysts can compare individual records with group-level statistics.


## Step 3: Implement the Concept


```python
df["City_Average_Sales"] = df.groupby("City")["Sales"].transform("mean")

print(df)
```

## Code Explanation

- `groupby("City")` creates groups based on City.
- `transform("mean")` calculates the average Sales for each city.
- The city average is placed back into every row belonging to that city.
- The number of rows remains the same as the original DataFrame.

### AI/ML Use

It can create group-level features such as customer average spending or regional average sales.

### Data Analytics Use

It helps compare individual values with the average of their group.

# 4. filter()

## Step 1 – Understand the Concept

`filter()` is used with grouped data to keep only groups that satisfy a specific condition.

The condition is evaluated for each group.

### Why is it needed?

It helps us focus on groups that meet a particular requirement instead of keeping every group.

## Step 2 – Demonstrate the Concept

### Simple Example

We can keep only cities whose total sales are greater than 70,000.

### Real-World Example

A company can keep only regions that have more than a certain number of customers.

### Business Example

A business can identify product categories whose total sales exceed a target.

### AI/ML Use Case

It can be used to retain groups with enough observations before creating group-based features.

### Data Analytics Use Case

Analysts can focus only on groups that satisfy a business condition.


## Step 3: Implement the Concept

```python
filtered_groups = df.groupby("City").filter(
    lambda group: group["Sales"].sum() > 70000
)

print(filtered_groups)
```

## Code Explanation

- `groupby("City")` creates groups based on City.
- `filter()` checks each city group.
- `lambda group: group["Sales"].sum() > 70000` keeps only cities whose total sales are greater than 70,000.
- All rows belonging to qualifying cities are retained.

### AI/ML Use

It can help remove groups with insufficient data before certain analysis or feature creation.

### Data Analytics Use

It helps focus analysis on groups that meet a specific business condition.

# 5. Pivot Tables

## Step 1 – Understand the Concept

A Pivot Table is used to summarize and reorganize data based on different categories.

Pandas provides `pivot_table()` to create Pivot Tables.

It can calculate values such as:

- Sum
- Mean
- Count
- Minimum
- Maximum

### Why is it needed?

Pivot Tables make it easier to compare multiple categories and summarize large datasets.

## Step 2 – Demonstrate the Concept

### Simple Example

We can create a Pivot Table showing total sales for each city and product.

### Real-World Example

A company can compare product sales across different regions.

### Business Example

A business manager can see which products perform best in each city.

### AI/ML Use Case

Pivot Tables can help explore relationships between categorical variables before feature engineering.

### Data Analytics Use Case

Analysts commonly use Pivot Tables to summarize and compare business data.


## Step 3: Implement the Concept

```python
pivot = pd.pivot_table(
    df,
    values="Sales",
    index="City",
    columns="Product",
    aggfunc="sum",
    fill_value=0
)

print(pivot)
```

## Code Explanation

- `pd.pivot_table()` creates a summarized table.
- `values="Sales"` specifies the values to summarize.
- `index="City"` places cities as rows.
- `columns="Product"` places products as columns.
- `aggfunc="sum"` calculates total sales.
- `fill_value=0` replaces missing combinations with 0.

### AI/ML Use

Pivot Tables can help understand relationships between categorical variables during exploratory analysis.

### Data Analytics Use

They provide an easy way to compare business metrics across multiple categories.

# 6. Crosstab

## Step 1 – Understand the Concept

A Crosstab, or cross-tabulation, is used to calculate the frequency or count of combinations between categorical variables.

Pandas provides `pd.crosstab()` for creating cross-tabulations.

### Why is it needed?

It helps us understand the relationship between two or more categorical variables.

## Step 2 – Demonstrate the Concept

### Simple Example

We can count how many customers belong to each combination of City and Product.

### Real-World Example

A company can study which products are purchased by customers from different cities.

### Business Example

A business can compare product preferences across different regions.

### AI/ML Use Case

Crosstab can help explore relationships between categorical features and identify patterns before model training.

### Data Analytics Use Case

Analysts can use it to understand frequency and relationships between categorical variables.


## Step 3: Implement the Concept

```python
cross_table = pd.crosstab(
    df["City"],
    df["Product"]
)

print(cross_table)
```

## Code Explanation

- `pd.crosstab()` creates a frequency table.
- `df["City"]` becomes the row categories.
- `df["Product"]` becomes the column categories.
- Each cell shows how many records belong to that City and Product combination.

### AI/ML Use

It helps explore relationships between categorical features before model development.

### Data Analytics Use

It helps identify category combinations and their frequencies.
