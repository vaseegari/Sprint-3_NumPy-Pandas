# 1. Rename Columns

## Step 1 – Understand the Concept

Renaming columns means changing the existing column names of a DataFrame.

Pandas provides the `rename()` method to change column names.

### Why is it needed?

Clear and meaningful column names make the dataset easier to understand, analyze, and use in further processing.

## Step 2 – Demonstrate the Concept

### Simple Example

A column named `Marks` can be renamed to `Score`.

### Real-World Example

A customer dataset may have `Cust_Name`, which can be renamed to `Customer_Name` for better readability.

### Business Example

A sales dataset can rename `Amt` to `Sales_Amount` so employees can understand the column clearly.

### AI/ML Use Case

Clear column names make it easier to identify features during machine learning preprocessing.

### Data Analytics Use Case

Analysts can rename unclear or inconsistent column names before performing analysis.


## Step 3: Implement the Concept

```python
import pandas as pd

df = pd.DataFrame({
    "Name": ["Arun", "Priya", "Kumar", "Anitha"],
    "Age": [22, 24, 23, 25],
    "Marks": [85, 90, 78, 88]
})

df = df.rename(columns={"Marks": "Score"})

print(df)
```

## Code Explanation

- `rename()` is used to change column names.
- `columns={"Marks": "Score"}` tells Pandas to change `Marks` to `Score`.
- The other column names remain unchanged.

### AI/ML Use

Meaningful feature names make preprocessing and model development easier.

### Data Analytics Use

Clear column names improve readability and reduce confusion during analysis.

# 2. Drop Rows

## Step 1 – Understand the Concept

Dropping rows means removing selected rows from a DataFrame.

Pandas provides the `drop()` method to remove rows using their index labels.

### Why is it needed?

Unwanted, incorrect, or unnecessary records can be removed before analysis.

## Step 2 – Demonstrate the Concept

### Simple Example

If one student record is incorrect, we can remove that row.

### Real-World Example

A customer dataset may contain a test or invalid customer record that needs to be removed.

### Business Example

A company can remove cancelled or invalid transactions from a dataset.

### AI/ML Use Case

Unwanted or invalid observations can be removed before training a machine learning model.

### Data Analytics Use Case

Analysts can remove irrelevant records before performing analysis.


## Step 3: Implement the Concept

```python
df = pd.DataFrame({
    "Name": ["Arun", "Priya", "Kumar", "Anitha"],
    "Age": [22, 24, 23, 25],
    "Score": [85, 90, 78, 88]
})

df = df.drop(index=2)

print(df)
```

## Code Explanation

- `drop(index=2)` removes the row whose index is `2`.
- The remaining rows are kept in the DataFrame.
- `drop()` is useful for removing unwanted records.

### AI/ML Use

It can remove invalid or irrelevant observations before model training.

### Data Analytics Use

It helps clean datasets by removing unwanted rows.

# 3. Drop Columns

## Step 1 – Understand the Concept

Dropping columns means removing one or more columns from a DataFrame.

Pandas provides the `drop()` method with `axis=1` or `columns=` for this purpose.

### Why is it needed?

Unnecessary columns can be removed to simplify the dataset and focus on useful information.

## Step 2 – Demonstrate the Concept

### Simple Example

If `Age` is not required for an analysis, we can remove the Age column.

### Business Example

### AI/ML Use Case

Unnecessary features can be removed during feature selection.

### Data Analytics Use Case

Analysts can remove irrelevant columns to make analysis simpler.


## Step 3: Implement the Concept

```python
df = pd.DataFrame({
    "Name": ["Arun", "Priya", "Kumar", "Anitha"],
    "Age": [22, 24, 23, 25],
    "Score": [85, 90, 78, 88]
})

df = df.drop(columns=["Age"])

print(df)
```

# 4. Insert Columns

## Step 1 – Understand the Concept

Inserting a column means adding a new column at a specific position in a DataFrame.

Pandas provides the `insert()` method to add a column at a particular location.

### Why is it needed?

It allows us to add calculated or new information exactly where we want it in the DataFrame.

## Step 2 – Demonstrate the Concept

### Simple Example

We can insert a new `Grade` column after the Name column.

### Real-World Example

A customer dataset can have a new `Customer_Type` column added based on customer information.

### Business Example

A sales dataset can add a `Profit` column based on Revenue and Cost.

### AI/ML Use Case

New features can be created and inserted into a dataset during feature engineering.

### Data Analytics Use Case

Analysts can add calculated columns for further analysis.


## Step 3: Implement the Concept

```python
df = pd.DataFrame({
    "Name": ["Arun", "Priya", "Kumar"],
    "Score": [85, 90, 78]
})

df.insert(1, "Grade", ["A", "A+", "B"])

print(df)
```

## Code Explanation

- `insert()` adds a new column at a specific position.
- `1` represents the position where the column should be inserted.
- `"Grade"` is the new column name.
- The list contains the values for the new column.

### AI/ML Use

It can be used during feature engineering to add new features.

### Data Analytics Use

It helps analysts add calculated or categorized information to a DataFrame.

# 5. assign()

## Step 1 – Understand the Concept

`assign()` is used to create new columns in a DataFrame.

It is especially useful when the new column is calculated from existing columns.

### Why is it needed?

It provides a convenient way to create new columns while keeping the original DataFrame unchanged unless the result is assigned back.

## Step 2 – Demonstrate the Concept

### Simple Example

We can calculate total marks from different subject marks.

### Real-World Example

A customer dataset can calculate total spending from multiple purchase columns.

### Business Example

A sales dataset can calculate Profit using Sales and Cost.

### AI/ML Use Case

`assign()` can be used to create new features during feature engineering.

### Data Analytics Use Case

Analysts can create calculated columns without manually changing the original data.


## Step 3: Implement the Concept

```python
df = pd.DataFrame({
    "Product": ["Laptop", "Phone", "Tablet"],
    "Sales": [100000, 75000, 45000],
    "Cost": [80000, 60000, 35000]
})

df = df.assign(Profit=df["Sales"] - df["Cost"])

print(df)
```

## Code Explanation

- `assign()` creates a new column called `Profit`.
- `df["Sales"] - df["Cost"]` calculates the profit.
- The result is assigned to the `Profit` column.
- The modified DataFrame is stored back in `df`.

### AI/ML Use

It can create new features from existing features during feature engineering.

### Data Analytics Use

It helps create calculated columns for business analysis.

# 6. apply()

## Step 1 – Understand the Concept

`apply()` is used to apply a function to values in a Pandas Series or across rows or columns of a DataFrame.

### Why is it needed?

It allows us to perform custom operations on data when a simple built-in operation is not enough.

## Step 2 – Demonstrate the Concept

### Simple Example

We can use `apply()` to increase each student's score by 5 marks.

### Real-World Example

A company can apply a custom transformation to customer data.

### Business Example

A business can calculate a discount based on the purchase amount.

### AI/ML Use Case

`apply()` can be used for custom feature transformations during preprocessing.

### Data Analytics Use Case

Analysts can apply custom functions to transform or clean data.


## Step 3: Implement the Concept

```python
df = pd.DataFrame({
    "Name": ["Arun", "Priya", "Kumar"],
    "Score": [85, 90, 78]
})

df["Updated_Score"] = df["Score"].apply(lambda x: x + 5)

print(df)
```

## Code Explanation

- `apply()` applies a function to every value in the Score column.
- `lambda x: x + 5` adds 5 to each score.
- A new column called `Updated_Score` stores the result.

### AI/ML Use

It can be used for custom feature transformations.

### Data Analytics Use

It helps perform custom calculations or transformations on data.

# 7. map()

## Step 1 – Understand the Concept

`map()` is used to transform or replace values in a Pandas Series using a function or dictionary.

### Why is it needed?

It is useful when we want to convert existing values into new values based on a defined mapping.

## Step 2 – Demonstrate the Concept

### Simple Example

We can convert short grade codes into full descriptions.

### Real-World Example

A dataset containing city codes can be converted into city names.

### Business Example

A company can convert product codes into product categories.

### AI/ML Use Case

`map()` can convert categorical values into meaningful or numerical representations during preprocessing.

### Data Analytics Use Case

Analysts can transform category labels into more meaningful values.


## Step 3: Implement the Concept

```python
df = pd.DataFrame({
    "Name": ["Arun", "Priya", "Kumar"],
    "Grade": ["A", "A+", "B"]
})

grade_map = {
    "A": "Good",
    "A+": "Excellent",
    "B": "Average"
}

df["Grade_Description"] = df["Grade"].map(grade_map)

print(df)
```

## Code Explanation

- `grade_map` contains the mapping between grade codes and descriptions.
- `map()` replaces each grade with its corresponding description.
- `"A"` becomes `"Good"`, `"A+"` becomes `"Excellent"`, and `"B"` becomes `"Average"`.

### AI/ML Use

It can be used to transform categorical values into useful representations during preprocessing.

### Data Analytics Use

It helps convert codes or categories into meaningful labels.

# 8. replace()

## Step 1 – Understand the Concept

`replace()` is used to replace existing values with new values in a DataFrame or Series.

### Why is it needed?

It is useful for correcting values, standardizing categories, and replacing unwanted values.

## Step 2 – Demonstrate the Concept

### Simple Example

We can replace `"M"` with `"Male"` and `"F"` with `"Female"`.

### Real-World Example

A customer dataset may contain different spellings such as `"Chennai"` and `"Chennnai"` that need correction.

### Business Example

A company can standardize product categories before generating reports.

### AI/ML Use Case

Inconsistent categorical values can be standardized before model training.

### Data Analytics Use Case

Analysts can correct inconsistent or incorrect values before analysis.


## Step 3: Implement the Concept

```python
df = pd.DataFrame({
    "Name": ["Arun", "Priya", "Kumar", "Anitha"],
    "Gender": ["M", "F", "M", "F"]
})

df["Gender"] = df["Gender"].replace({
    "M": "Male",
    "F": "Female"
})

print(df)
```

## Code Explanation

- `replace()` changes existing values into new values.
- `"M"` is replaced with `"Male"`.
- `"F"` is replaced with `"Female"`.
- This helps standardize the values in the Gender column.

### AI/ML Use

Standardized categories improve the quality of data used for machine learning.

### Data Analytics Use

It helps clean and standardize inconsistent values before analysis.

# 9. Duplicate Handling

## Step 1 – Understand the Concept

Duplicate handling means identifying and removing repeated records from a DataFrame.

Pandas provides:
- `duplicated()` to identify duplicate rows
- `drop_duplicates()` to remove duplicate rows

### Why is it needed?

Duplicate records can affect calculations, analysis, and machine learning results.
Removing unnecessary duplicates improves data quality.

## Step 2 – Demonstrate the Concept

### Simple Example

If the same student record appears twice, one duplicate record can be removed.

### Real-World Example

A customer may accidentally be registered more than once in a database.

### Business Example

A sales transaction may be recorded twice, which can incorrectly increase reported revenue.

### AI/ML Use Case

Duplicate records can cause certain observations to have too much influence during model training.

### Data Analytics Use Case

Removing duplicates prevents incorrect counts, totals, and statistical analysis.


## Step 3: Implement the Concept

```python
df = pd.DataFrame({
    "Name": ["Arun", "Priya", "Kumar", "Arun"],
    "Score": [85, 90, 78, 85]
})

print("Duplicate rows:")
print(df[df.duplicated()])

df = df.drop_duplicates()

print("\nData after removing duplicates:")
print(df)
```

## Code Explanation

- `duplicated()` identifies rows that are duplicates of previous rows.
- `df[df.duplicated()]` displays the duplicate records.
- `drop_duplicates()` removes duplicate rows.
- The cleaned DataFrame contains only unique records.

### AI/ML Use

Removing duplicates prevents repeated observations from unnecessarily influencing the model.

### Data Analytics Use

It prevents duplicate records from affecting counts, totals, averages, and other calculations.
