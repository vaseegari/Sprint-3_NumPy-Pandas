# 1. loc

## Step 1 – Understand the Concept

`loc` is used to select rows and columns from a Pandas DataFrame using labels or column names.

We can use `loc` to select:
- A specific row
- Multiple rows
- Specific columns
- Rows based on conditions

### Why is it needed?

`loc` makes it easy to select specific data using row labels and column names.

## Step 2 – Demonstrate the Concept

### Simple Example

If a DataFrame contains student names and marks, `loc` can select a student's row or specific columns.

### Real-World Example

A customer dataset can use `loc` to select the records of specific customers.

### Business Example

A company can select the sales information for a particular product or region.

### AI/ML Use Case

`loc` can be used to select specific features or rows while preparing training data.

### Data Analytics Use Case

Analysts can select required rows and columns for detailed analysis.


## Step 3: Implement the Concept

```python
import pandas as pd

df = pd.DataFrame({
    "Name": ["Arun", "Priya", "Kumar", "Anitha", "Rahul"],
    "Age": [22, 24, 23, 25, 21],
    "Marks": [85, 90, 78, 88, 76],
    "City": ["Chennai", "Mumbai", "Chennai", "Delhi", "Bangalore"]
})

print(df.loc[0])
print()
print(df.loc[0:2, ["Name", "Marks"]])
```

## Code Explanation

- `df.loc[0]` selects the row with label `0`.
- `df.loc[0:2, ["Name", "Marks"]]` selects rows 0 to 2 and only the Name and Marks columns.
- `loc` uses labels and column names.

### AI/ML Use

It can be used to select required observations and features during data preparation.

### Data Analytics Use

It helps analysts select specific rows and columns for analysis.

# 2. iloc

## Step 1 – Understand the Concept

`iloc` is used to select rows and columns using their integer positions.

The position starts from `0`.

### Why is it needed?

`iloc` is useful when we want to select data based on its numerical position instead of labels or column names.

## Step 2 – Demonstrate the Concept

### Simple Example

The first row has position `0`, and the second row has position `1`.

### Real-World Example

We can select the first few records from a customer dataset using their positions.

### Business Example

A business can select specific columns from a sales dataset based on their positions.

### AI/ML Use Case

`iloc` can be used to separate specific rows or columns during data preprocessing.

### Data Analytics Use Case

Analysts can quickly select a range of rows or columns using their positions.


## Step 3: Implement the Concept

```python
print(df.iloc[0])
print()
print(df.iloc[0:3, 0:2])
```

## Code Explanation

- `df.iloc[0]` selects the first row because its position is `0`.
- `df.iloc[0:3, 0:2]` selects the first 3 rows and first 2 columns.
- `iloc` works with integer positions.

### AI/ML Use

It can be used to select specific portions of a dataset during preprocessing.

### Data Analytics Use

It helps analysts select data based on row and column positions.

# 3. Boolean Indexing

## Step 1 – Understand the Concept

Boolean indexing is used to select rows based on a condition.

The condition produces `True` or `False` values.

Only rows where the condition is `True` are selected.

### Why is it needed?

It allows us to filter data based on specific conditions.

## Step 2 – Demonstrate the Concept

### Simple Example

We can select students whose marks are greater than 80.

### Real-World Example

A company can select customers whose purchase amount is greater than ₹10,000.

### Business Example

A sales team can identify products with sales above a particular target.

### AI/ML Use Case

Boolean indexing can be used to remove unwanted records or select observations that satisfy certain conditions.

### Data Analytics Use Case

Analysts use conditions to filter datasets and focus on relevant records.


## Step 3: Implement the Concept

```python
high_marks = df[df["Marks"] > 80]

print(high_marks)
```

## Code Explanation

- `df["Marks"] > 80` creates a Boolean condition.
- Rows with marks greater than 80 become `True`.
- `df[...]` selects only the rows where the condition is `True`.

### AI/ML Use

It can be used to select relevant observations and remove unwanted records during preprocessing.

### Data Analytics Use

It helps analysts focus on records that satisfy a particular condition.

# 4. Filtering Rows

## Step 1 – Understand the Concept

Filtering rows means selecting only the rows that satisfy a specific condition.

We can filter rows based on numerical, text, or multiple conditions.

### Why is it needed?

Filtering helps us work only with the data that is relevant to a particular analysis.

## Step 2 – Demonstrate the Concept

### Simple Example

Select students whose age is greater than 22.

### Real-World Example

Select customers from Chennai.

### Business Example

Select products whose sales are above ₹50,000.

### AI/ML Use Case

Filter training data based on specific requirements before model training.

### Data Analytics Use Case

Filter records to perform focused analysis on a particular group.


## Step 3: Implement the Concept

```python
filtered_rows = df[df["Age"] > 22]

print(filtered_rows)
```

## Code Explanation

- `df["Age"] > 22` checks the age condition.
- Only rows where Age is greater than 22 are selected.
- The result is stored in `filtered_rows`.

### AI/ML Use

Filtering can help prepare a relevant subset of data for model development.

### Data Analytics Use

It helps analysts study a specific group of records.

# 5. query()

## Step 1 – Understand the Concept

`query()` is used to filter rows using a string-based condition.

It provides a simple and readable way to write filtering conditions.

### Why is it needed?

For complex filtering conditions, `query()` can make the code easier to read.

## Step 2 – Demonstrate the Concept

### Simple Example

We can select students whose marks are greater than 80 using `query()`.

### Real-World Example

A company can select customers whose age is above 25.

### Business Example

A business can find products where sales are greater than 50,000.

### AI/ML Use Case

`query()` can be used to create a filtered dataset before machine learning preprocessing.

### Data Analytics Use Case

Analysts can use readable conditions to filter large datasets.


## Step 3: Implement the Concept

- The condition is written as a string.
- `query()` makes filtering conditions easy to read.

### AI/ML Use

It can be used to create subsets of data during preprocessing.

### Data Analytics Use

It helps analysts write readable filtering conditions.

# 6. isin()

## Step 1 – Understand the Concept

`isin()` checks whether values belong to a given list of values.

It returns `True` when a value is present in the specified list.

### Why is it needed?

It is useful when we want to filter data based on multiple possible values.

## Step 2 – Demonstrate the Concept

### Simple Example

We can select students from Chennai or Delhi.

### Real-World Example

A company can select customers from selected cities.

### Business Example

A business can analyze sales for selected products.

### AI/ML Use Case

`isin()` can be used to select observations belonging to specific categories.

### Data Analytics Use Case

Analysts can filter data for multiple categories at the same time.


## Step 3: Implement the Concept

```python
selected_cities = df[df["City"].isin(["Chennai", "Delhi"])]

print(selected_cities)
```

## Code Explanation

- `isin(["Chennai", "Delhi"])` checks whether each city belongs to the given list.
- Rows matching either Chennai or Delhi are selected.
- It is useful when there are multiple acceptable values.

### AI/ML Use

It can be used to select specific categories for preprocessing or analysis.

### Data Analytics Use

It helps analysts filter multiple categories efficiently.

# 7. between()

## Step 1 – Understand the Concept

`between()` checks whether values fall within a specified range.

By default, the boundary values are included.

### Why is it needed?

It makes range-based filtering simple and readable.

## Step 2 – Demonstrate the Concept

### Simple Example

We can select students whose marks are between 80 and 90.

### Real-World Example

A company can select customers whose age is between 20 and 30.

### Business Example

A business can identify products with prices within a specific price range.

### AI/ML Use Case

It can be used to select observations within a specific numerical range during preprocessing.

### Data Analytics Use Case

Analysts can use it to study a particular range of values.


## Step 3: Implement the Concept

```python
marks_range = df[df["Marks"].between(80, 90)]

print(marks_range)
```

## Code Explanation

- `between(80, 90)` checks whether Marks is between 80 and 90.
- By default, both 80 and 90 are included.
- Only rows satisfying the range condition are returned.

### AI/ML Use

It can help select observations within a specific feature range.

### Data Analytics Use

It helps analyze data within a defined numerical range.

# 8. Sorting Data

## Step 1 – Understand the Concept

Sorting means arranging data in a specific order.

Pandas provides `sort_values()` to sort a DataFrame based on one or more columns.

We can sort in:
- Ascending order
- Descending order

### Why is it needed?

Sorting helps us easily identify the highest, lowest, earliest, latest, or otherwise ordered records.

## Step 2 – Demonstrate the Concept

### Simple Example

We can arrange students from the lowest marks to the highest marks.

### Real-World Example

A company can sort customers based on purchase amount.

### Business Example

A sales team can sort products from highest sales to lowest sales to identify top-performing products.

### AI/ML Use Case

Sorting can help inspect model-related metrics or organize data during exploratory analysis.

### Data Analytics Use Case

Analysts can sort data to identify top and bottom values.


## Step 3: Implement the Concept

```python
sorted_df = df.sort_values("Marks", ascending=False)

print(sorted_df)
```

## Code Explanation

- `sort_values("Marks")` sorts the DataFrame using the Marks column.
- `ascending=False` sorts the values from highest to lowest.
- If `ascending=True`, the values are sorted from lowest to highest.

### AI/ML Use

Sorting can help inspect model results, feature values, or evaluation metrics during analysis.

### Data Analytics Use

It helps identify highest and lowest values and rank business records.
