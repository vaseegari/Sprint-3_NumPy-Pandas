# 1. What is NumPy?

## Step 1 – Understand the Concept

NumPy stands for Numerical Python. It is a Python library used to work with numerical data and perform mathematical operations efficiently.

The main object in NumPy is the `ndarray`, which is used to store multiple values in an array.

NumPy is commonly used in Data Analytics, Scientific Computing, and Machine Learning.

## Step 2 – Demonstrate the Concept

Suppose we have the marks of five students:

80, 75, 90, 85, 70

# 4. Creating Arrays

## Step 1 – Understand the Concept

NumPy provides different functions for creating arrays.

Common functions are:

- `np.ones()` – creates an array containing ones
- `np.arange()` – creates values within a range
- `np.linspace()` – creates evenly spaced values

## Step 2 – Demonstrate the Concept

# 5. Dimensions

## Step 1 – Understand the Concept

The dimension of an array tells us how many levels of data the array contains.

Examples:

- 0D → single value
- 1D → one-dimensional sequence
- 2D → rows and columns
- 3D → collection of 2D arrays

NumPy provides the `.ndim` attribute to find the number of dimensions.

## Step 2 – Demonstrate the Concept

### Simple Example

A single number is 0D.

A list of marks is 1D.

A table containing marks is 2D.

### Real-World Example

A school can represent one student's marks as 1D and the marks of many students and subjects as 2D.

### Business Example

Monthly sales for different products can be represented as a 2D array.

### AI/ML Use Case

An image can be represented as a 2D array of pixels. A color image can be represented using a 3D array.

### Data Analytics Use Case

Understanding dimensions helps analysts correctly organize and process datasets.


## Step 3 – Implement the Concept (Python)

````python
import numpy as np

array_0d = np.array(10)
array_1d = np.array([10, 20, 30])
array_2d = np.array([[10, 20], [30, 40]])
array_3d = np.array([[[1, 2], [3, 4]], [[5, 6], [7, 8]]])

print("0D:", array_0d.ndim)
print("1D:", array_1d.ndim)
print("2D:", array_2d.ndim)
print("3D:", array_3d.ndim)
````

## Code Explanation

### Why `.ndim` Is Used

```text

# 11. Reshaping

## Step 1 – Understand the Concept

Reshaping changes the structure of an array without changing its values.

The total number of elements must remain the same.

For example, an array with 6 elements can be reshaped into:

- `(2, 3)`
- `(3, 2)`

## Step 2 – Demonstrate the Concept

### Simple Example

Original:

[1, 2, 3, 4, 5, 6]

After reshaping to `(2, 3)`:

[1, 2, 3]
[4, 5, 6]

### Real-World Example

A list of student marks can be reshaped into rows representing students and columns representing subjects.

Monthly sales values can be reshaped into a table where rows represent months and columns represent products.

### AI/ML Use Case

Reshaping is often required to convert data into the input shape expected by a Machine Learning or Deep Learning model.

### Data Analytics Use Case

Reshaping helps organize numerical data into a useful row-and-column structure.


## Step 3 – Implement the Concept (Python)

````python
import numpy as np

numbers = np.array([1, 2, 3, 4, 5, 6])

reshaped = numbers.reshape(2, 3)

print("Original:", numbers)
print("Reshaped:")
print(reshaped)
````

`reshape()` is used when we need to change the structure of an array while keeping the same values.

# 13. Copy vs View

## Step 1 – Understand the Concept

NumPy provides two ways to create another array from an existing array.

### Copy

A copy creates an independent array.

Changes made to the copy do not affect the original array.

### View

A view shares the same underlying data.

Changes made to the view can affect the original array

## Step 2 – Demonstrate the Concept

### Simple Example

Suppose we have an array containing student marks.

If we create a copy and change it, the original marks remain unchanged.

If we create a view and change it, the original array can also change.

### Real-World Example

A teacher may create a separate copy of marks before modifying them for analysis.

### Business Example

A company may create a copy of sales data before making changes, so that the original sales data remains safe.

### AI/ML Use Case

Understanding copy and view is important during preprocessing because accidental changes to original training data can affect model results.

### Data Analytics Use Case

Copies can be used when we need an independent dataset, while views can be useful when working with the same underlying data without creating another full copy.


## Step 3 – Implement the Concept (Python)

````python
import numpy as np

original = np.array([10, 20, 30])

copy_array = original.copy()
view_array = original.view()

copy_array[0] = 100
view_array[1] = 200

print("Original:", original)
print("Copy:", copy_array)
print("View:", view_array)
````

## Code Explanation

### What the Code Does

The code creates one copy and one view from the original array and modifies both.

### Why `copy()` and `view()` Are Used

- `copy()` creates independent data.
- `view()` creates another array that shares the same underlying data.

### Expected Output

```text
Original: [ 10 200  30]
Copy: [100  20  30]
View: [ 10 200  30]

# 14. Broadcasting

Broadcasting allows NumPy to perform operations between arrays of compatible shapes.

It can automatically apply a single value to every element of an array.

This avoids writing manual loops for simple numerical operations.

# 16. Universal Functions (ufuncs)

## Step 1 – Understand the Concept

Universal functions, called **ufuncs**, are NumPy functions that perform mathematical operations efficiently on array elements.

Examples include:

- `np.sqrt()` → square root
- `np.square()` → square
- `np.abs()` → absolute value
- `np.exp()` → exponential
- `np.log()` → logarithm
- `np.sin()` → sine
- `np.cos()` → cosine

ufuncs can perform operations on complete arrays without writing a normal Python loop.

## Step 2 – Demonstrate the Concept

### Simple Example

For:

[1, 4, 9, 16]

We can use `np.sqrt()` to calculate the square root of every value.

Result:

[1, 2, 3, 4]

### Real-World Example

A scientific application may calculate square roots, logarithms, or other mathematical values for thousands of measurements.

### Business Example

A business analyst may apply mathematical transformations to sales, revenue, or customer data before analysis.

### AI/ML Use Case

Mathematical transformations such as logarithm, exponential, and square root are commonly used in feature engineering and model calculations.

### Data Analytics Use Case

ufuncs can efficiently transform large numerical datasets before visualization or statistical analysis.


## Step 3 – Implement the Concept (Python)

````python
import numpy as np

numbers = np.array([1, 4, 9, 16])

square_root = np.sqrt(numbers)
square = np.square(numbers)
absolute = np.abs(np.array([-5, -10, 15]))

print("Square root:", square_root)
print("Square:", square)
print("Absolute:", absolute)
````

## Code Explanation

### What the Code Does

The code applies different NumPy mathematical functions to numerical arrays.

### Why the Functions Are Used

- `np.sqrt()` calculates the square root of each element.
- `np.square()` calculates the square of each element.
- `np.abs()` returns the positive magnitude of each element.

### Expected Output

```text
Square root: [1. 2. 3. 4.]
Square: [  1  16  81 256]
Absolute: [ 5 10 15]
