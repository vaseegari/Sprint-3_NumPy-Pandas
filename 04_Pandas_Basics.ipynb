# 1. What is Pandas?

## Step 1 – Understand the Concept

Pandas is a Python library used for working with structured and tabular data.

It provides two important data structures:

- Series
- DataFrame

Pandas makes it easier to read, clean, analyze, and manipulate data.

### Why do we need Pandas?

- To work with tabular data.
- To read data from CSV and other files.
- To clean and prepare data.
- To analyze datasets.
- To prepare data for AI/ML models.

## Step 2 – Demonstrate the Concept

### Simple Example

We can use Pandas to store student names and marks in a table.

### Real-World Example

A school can use Pandas to analyze student attendance and marks.

### Business Example

A company can use Pandas to analyze sales, customers, and product data.

### AI/ML Use Case

Pandas is commonly used during data preprocessing before training a machine learning model.

### Data Analytics Use Case

Pandas is used to filter, summarize, group, and analyze business datasets.


## Step 3 – Implement the Concept (Python)

```python
import pandas as pd

students = pd.DataFrame({
    "Name": ["Arun", "Priya", "Kumar"],
    "Marks": [85, 90, 78]
})

print(students)
```

## Code Explanation

- `import pandas as pd` imports the Pandas library.
- `pd.DataFrame()` creates a table-like data structure.
- The output contains student names and their marks.

### AI/ML Use

Pandas is used to load and preprocess datasets before giving them to machine learning models.

### Data Analytics Use

Pandas is used to analyze structured business and customer data.

# 2. Why Pandas?

## Step 1 – Understand the Concept

Pandas provides simple and powerful tools for handling structured data.

It allows us to easily perform operations such as:

- Filtering data
- Sorting data
- Selecting columns
- Handling missing values
- Grouping data
- Reading and writing files

### Why do we need Pandas?

Python lists and dictionaries can store data, but Pandas provides specialized tools for analyzing large tabular datasets efficiently.

## Step 2 – Demonstrate the Concept

### Simple Example

We can filter students whose marks are greater than 80.

### Real-World Example

A school can find students who scored above a particular mark.

### Business Example

A company can find products whose sales are greater than a specific amount.

### AI/ML Use Case

Pandas can be used to select relevant features and prepare training data.

### Data Analytics Use Case

Analysts can filter and summarize large datasets to find useful information.


## Step 3 – Implement the Concept (Python)

```python
import pandas as pd

data = pd.DataFrame({
    "Name": ["Arun", "Priya", "Kumar", "Divya"],
    "Marks": [85, 90, 65, 95]
})

high_scores = data[data["Marks"] > 80]

print(high_scores)
```

## Code Explanation

- `data["Marks"]` selects the Marks column.
- `data["Marks"] > 80` creates a condition.
- The condition is used to filter students with marks greater than 80.
- The output contains only students who scored above 80.

### AI/ML Use

Filtering can be used to select relevant records or features during preprocessing.

### Data Analytics Use

Filtering helps analysts focus on specific records based on business conditions.

A Series is useful for analyzing individual columns in a dataset.

# 6. Data Types

## Step 1 – Understand the Concept

Data types describe what kind of values are stored in a column.

Common Pandas data types include:

- `int64` – integer values
- `float64` – decimal values
- `object` – commonly used for text
- `bool` – True or False values

Pandas uses data types to understand and process the values correctly.

### Why do we need Data Types?

- To understand the structure of data.
- To perform correct calculations.
- To identify incorrect data.
- To prepare data for machine learning.

## Step 2 – Demonstrate the Concept

### Simple Example

Create a DataFrame containing names, ages, salaries, and active status.

### Real-World Example

Employee data contains text, integer, decimal, and Boolean values.

### Business Example

A company may have product names, prices, quantities, and availability.

### AI/ML Use Case

Correct data types are important during preprocessing and feature engineering.

### Data Analytics Use Case

Checking data types helps analysts identify whether columns are suitable for calculations or need conversion.


## Step 3 – Implement the Concept (Python)

```python
import pandas as pd

df = pd.DataFrame({
    "Name": ["Arun", "Priya", "Kumar"],
    "Age": [22, 24, 23],
    "Salary": [25000.50, 32000.75, 28000.00],
    "Active": [True, False, True]
})

print(df)
print("\nData Types:")
print(df.dtypes)
```

## Code Explanation

- `df.dtypes` returns the data type of each column.
- `Age` is stored as an integer.
- `Salary` is stored as a floating-point number.
- `Name` contains text.
- `Active` contains Boolean values.

### AI/ML Use

Correct data types help during feature preprocessing and model training.

### Data Analytics Use

Data types help analysts identify columns that can be used for calculations and analysis.

# 7. Index

## Step 1 – Understand the Concept

An index identifies each row in a Pandas Series or DataFrame.

By default, Pandas creates an integer index starting from 0.

We can also create a custom index.

### Why do we need Index?

- To identify rows.
- To access specific data.
- To organize data.
- To use meaningful labels.

## Step 2 – Demonstrate the Concept

### Simple Example

Create a DataFrame with a custom index representing employee IDs.

### Real-World Example

Student IDs can be used as row labels.

### Business Example

Customer IDs can be used as an index for customer records.

### AI/ML Use Case

Indexes help identify and track individual observations in datasets.

### Data Analytics Use Case

Indexes make it easier to access and organize rows during analysis.


## Step 3 – Implement the Concept (Python)

```python
import pandas as pd

df = pd.DataFrame({
    "Name": ["Arun", "Priya", "Kumar"],
    "Marks": [85, 90, 78]
}, index=["S101", "S102", "S103"])

print(df)
print("\nIndex:")
print(df.index)
```

## Code Explanation

- The `index` parameter assigns custom labels to rows.
- `S101`, `S102`, and `S103` identify the rows.
- `df.index` displays the index values.

### AI/ML Use

Indexes can help track individual records during preprocessing and analysis.

### Data Analytics Use

Meaningful indexes can make data easier to identify and organize.

# 9. Reading Data

## Step 1 – Understand the Concept

Pandas can read data from external files and convert it into a DataFrame.

One of the most commonly used functions is:

`pd.read_csv()`

It is used to read CSV files.

### Why do we need Reading Data?

Real-world datasets are usually stored in files or databases.

Reading data allows us to bring external data into Python for analysis and machine learning.

## Step 2 – Demonstrate the Concept

### Simple Example

Read a CSV file containing student information.

### Real-World Example

A school can read student records stored in a CSV file.

### Business Example

A company can read sales data stored in a CSV file.

### AI/ML Use Case

Machine learning datasets can be loaded into Pandas before preprocessing.

### Data Analytics Use Case

Analysts can load CSV files and perform data analysis using Pandas.


## Step 3 – Implement the Concept (Python)

```python
import pandas as pd

data = {
    "Name": ["Arun", "Priya", "Kumar"],
    "Marks": [85, 90, 78]
}

df = pd.DataFrame(data)

df.to_csv("students.csv", index=False)

print("students.csv file created successfully.")
```

```python
import pandas as pd

df = pd.read_csv("students.csv")

print(df)
```

## Code Explanation

- `pd.read_csv()` reads a CSV file.
- `"students.csv"` is the file name.
- The data is loaded into the DataFrame `df`.
- `df.head()` displays the first five rows.

### Expected Output

The first five records from the CSV file will be displayed.

### AI/ML Use

Datasets can be loaded before cleaning, preprocessing, and model training.

### Data Analytics Use

CSV files containing business data can be loaded and analyzed using Pandas.

# 10. Writing Data

## Step 1 – Understand the Concept

Writing data means saving a Pandas DataFrame into an external file.

For example, `to_csv()` can save a DataFrame as a CSV file.

### Why do we need Writing Data?

- To save processed data.
- To share datasets.
- To store analysis results.
- To create files for further processing.
- To save machine learning preprocessing results.

## Step 2 – Demonstrate the Concept

### Simple Example

Save a DataFrame into a CSV file.

### Real-World Example

A school can save updated student records to a CSV file.

### Business Example

A company can save processed sales data for reporting.

### AI/ML Use Case

Cleaned and processed datasets can be saved for later machine learning experiments.

### Data Analytics Use Case

Analysis results or filtered datasets can be exported for reports and further analysis.


## Step 3 – Implement the Concept (Python)

```python
import pandas as pd

df = pd.DataFrame({
    "Name": ["Arun", "Priya", "Kumar"],
    "Marks": [85, 90, 78]
})

df.to_csv("students_output.csv", index=False)

print("Data saved successfully.")
```

## Code Explanation

- `df.to_csv()` saves the DataFrame as a CSV file.
- `"students_output.csv"` is the output file name.
- `index=False` prevents Pandas from saving the DataFrame index as an extra column.
- The file will be created in the current working directory.

### Expected Output

A file named `students_output.csv` will be created.

### AI/ML Use

Processed datasets can be saved and reused for model training.

### Data Analytics Use

Filtered or summarized data can be exported for reporting and sharing.
