# 1. concat()

## Step 1 – Understand the Concept

`concat()` is used to combine two or more Pandas DataFrames along rows or columns.

It can combine datasets:
- Vertically (row-wise)
- Horizontally (column-wise)

### Why is it needed?

When data is stored in multiple DataFrames or files, `concat()` helps combine them into a single DataFrame.

## Step 2 – Demonstrate the Concept

### Simple Example

Two DataFrames containing January and February sales can be combined into one DataFrame.

### Real-World Example

A company may store monthly sales data in separate files. `concat()` can combine them into one dataset.

### Business Example

A business can combine sales records from different branches into one DataFrame.

### AI/ML Use Case

Data from different batches or sources can be combined before preprocessing and model training.

### Data Analytics Use Case

Analysts can combine datasets from multiple months, regions, or departments for analysis.


## Step 3: Implement the Concept

```python
import pandas as pd

jan_sales = pd.DataFrame({
    "Product": ["Laptop", "Phone"],
    "Sales": [50000, 30000]
})

feb_sales = pd.DataFrame({
    "Product": ["Tablet", "Monitor"],
    "Sales": [20000, 25000]
})

combined_sales = pd.concat(
    [jan_sales, feb_sales],
    ignore_index=True
)

print(combined_sales)
```

## Code Explanation

- `pd.concat()` combines multiple DataFrames.
- The DataFrames are combined row-wise by default.
- `ignore_index=True` creates a new continuous index.
- The result contains records from both January and February.

### AI/ML Use

It can combine data from different batches before model training.

### Data Analytics Use

It helps combine monthly, regional, or departmental datasets for analysis.

# 2. merge()

## Step 1 – Understand the Concept

`merge()` is used to combine two DataFrames based on one or more common columns.

It works similarly to SQL joins.

Common types of merge include:
- Inner join
- Left join
- Right join
- Outer join

### Why is it needed?

It allows us to combine related information stored in different datasets using a common key.


## Step 2 – Demonstrate the Concept

### Simple Example

One DataFrame contains customer information and another contains purchase information. Both have `Customer_ID`.

### Real-World Example

A company can combine customer details with their transaction history.

### Business Example

A business can combine product information with sales information using `Product_ID`.

### AI/ML Use Case

Different feature datasets can be merged using a common identifier before model training.

### Data Analytics Use Case

Analysts can combine related datasets to perform more complete analysis.


## Step 3: Implement the Concept

```python
customers = pd.DataFrame({
    "Customer_ID": [101, 102, 103],
    "Name": ["Arun", "Priya", "Kumar"]
})

orders = pd.DataFrame({
    "Customer_ID": [101, 102, 104],
    "Amount": [5000, 7000, 3000]
})

merged_data = pd.merge(
    customers,
    orders,
    on="Customer_ID",
    how="inner"
)

print(merged_data)
```

## Code Explanation

- `pd.merge()` combines two DataFrames using a common column.
- `on="Customer_ID"` specifies the common key.
- `how="inner"` keeps only Customer_ID values available in both DataFrames.
- Customer 101 and 102 appear in the final result because they exist in both datasets.

### AI/ML Use

It can combine different datasets containing features related to the same observation.

### Data Analytics Use

It helps combine customer, transaction, product, and other related business data.

# 3. join()

## Step 1 – Understand the Concept

`join()` is used to combine DataFrames based primarily on their indexes.

It is useful when two DataFrames have related information with matching index values.

### Why is it needed?

It provides a convenient way to combine columns from DataFrames using their indexes.

## Step 2 – Demonstrate the Concept

### Simple Example

One DataFrame contains student names and another contains their marks. Both use the same student index.

### Real-World Example

A company may have employee details and employee performance data stored separately with the same employee ID as the index.

### Business Example

A business can combine product details and inventory information using matching indexes.

### AI/ML Use Case

`join()` can combine feature DataFrames that share the same observation index.

### Data Analytics Use Case

Analysts can combine related datasets when their indexes represent the same entities.


## Step 3: Implement the Concept

```python
student_info = pd.DataFrame({
    "Name": ["Arun", "Priya", "Kumar"]
}, index=[101, 102, 103])

student_marks = pd.DataFrame({
    "Marks": [85, 90, 78]
}, index=[101, 102, 103])

joined_data = student_info.join(student_marks)

print(joined_data)
```

## Code Explanation

- `join()` combines the two DataFrames using their indexes.
- Both DataFrames use student IDs as their indexes.
- The Name and Marks columns are combined into one DataFrame.

### AI/ML Use

It can combine feature sets that have the same observation index.

### Data Analytics Use

It helps combine related information when the DataFrames share a common index.

# 4. append()

## Step 1 – Understand the Concept

`append()` was previously used to add rows from one Pandas DataFrame to another.

However, the DataFrame `append()` method was deprecated in Pandas 1.4 and removed in Pandas 2.0.

Therefore, we should NOT use:

`df.append(other_df)`

in modern Pandas versions.

### Recommended Alternative

Use `pd.concat()` instead.

### Why is it needed?

`pd.concat()` provides the modern and supported way to combine DataFrames row-wise.

## Step 2 – Demonstrate the Concept

### Simple Example

Older Pandas code could use `append()` to add new rows to a DataFrame.

### Real-World Example

A company may receive new customer records and need to add them to an existing dataset.

### Business Example

A business can combine new daily sales records with existing sales data using `concat()`.

### AI/ML Use Case

New batches of training data can be combined with existing data using `concat()`.

### Data Analytics Use Case

Analysts can add records from different periods using `concat()`.

### Important Note

`append()` is deprecated and removed from modern Pandas.

Use:

`pd.concat([df1, df2], ignore_index=True)`

instead.


## Step 3: Implement the Concept

```python
old_data = pd.DataFrame({
    "Name": ["Arun", "Priya"],
    "Marks": [85, 90]
})

new_data = pd.DataFrame({
    "Name": ["Kumar", "Anitha"],
    "Marks": [78, 88]
})

# Modern alternative to append()
combined_data = pd.concat(
    [old_data, new_data],
    ignore_index=True
)

print(combined_data)
```

## Code Explanation

- `append()` is not used because it has been deprecated and removed from modern Pandas.
- `pd.concat()` is the recommended alternative.
- `[old_data, new_data]` provides the DataFrames to combine.
- `ignore_index=True` creates a new continuous index.

### AI/ML Use

It can combine new data batches with existing training data.

### Data Analytics Use

It can combine records collected from different time periods or sources.

### Important Interview Point

**Question:** What is the alternative to the deprecated `DataFrame.append()` method?

**Answer:** Use `pd.concat()` to combine DataFrames.

# 5. combine_first()

## Step 1 – Understand the Concept

`combine_first()` combines two DataFrames by filling missing values in the first DataFrame with values from the second DataFrame.

The values already present in the first DataFrame are preserved.

### Why is it needed?

It is useful when we have two datasets containing complementary information and want to fill missing values using another dataset.

## Step 2 – Demonstrate the Concept

### Simple Example

If a student's marks are missing in one DataFrame but available in another, `combine_first()` can fill the missing mark.

### Real-World Example

A customer database may have missing phone numbers that are available in another data source.

### Business Example

A company can fill missing product prices using information from another dataset.

### AI/ML Use Case

It can help combine complementary feature data and fill missing values before model training.

### Data Analytics Use Case

Analysts can combine datasets to fill missing information without overwriting existing valid values.


## Step 3: Implement the Concept

```python
data1 = pd.DataFrame({
    "Name": ["Arun", "Priya", "Kumar"],
    "Marks": [85, None, 78]
}, index=[101, 102, 103])

data2 = pd.DataFrame({
    "Name": ["Arun", "Priya", "Kumar"],
    "Marks": [80, 90, 75]
}, index=[101, 102, 103])

combined_data = data1.combine_first(data2)

print(combined_data)
```

## Code Explanation

- `combine_first()` checks the first DataFrame for missing values.
- If a value is missing, it takes the corresponding value from the second DataFrame.
- Arun's mark remains `85` because it already exists in the first DataFrame.
- Priya's missing mark is filled with `90` from the second DataFrame.
- Kumar's mark remains `78`.

### AI/ML Use

It can help combine complementary feature datasets and fill missing feature values.

### Data Analytics Use

It helps recover missing information from another compatible data source.
