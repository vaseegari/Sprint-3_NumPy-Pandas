# 1. CSV Files

## Step 1 – Understand the Concept

CSV stands for Comma-Separated Values.

A CSV file stores data in rows and columns, where values are usually separated by commas.

CSV is one of the most commonly used formats for storing and sharing tabular data.

Pandas provides:

- `pd.read_csv()` – to read CSV files
- `df.to_csv()` – to write CSV files

### Why do we need CSV Files?

- Easy to create and share.
- Supported by many applications.
- Useful for storing datasets.
- Commonly used in Data Analytics and AI/ML.

## Step 2 – Demonstrate the Concept

### Simple Example

Create a DataFrame and save it as a CSV file, then read it again.

### Real-World Example

A school can store student records in a CSV file.

### Business Example

A company can store daily sales data in a CSV file.

### AI/ML Use Case

Machine learning datasets are often stored as CSV files and loaded using Pandas.

### Data Analytics Use Case

Analysts can read CSV files to calculate sales, averages, trends, and other metrics.


## Step 3 – Implement the Concept (Python)

```python
import pandas as pd

data = {
    "Name": ["Arun", "Priya", "Kumar"],
    "Marks": [85, 90, 78]
}

df = pd.DataFrame(data)

# Write data to CSV
df.to_csv("students.csv", index=False)

# Read data from CSV
df_read = pd.read_csv("students.csv")

print(df_read)
```

## Code Explanation

- `pd.DataFrame()` creates a DataFrame.
- `df.to_csv()` writes the DataFrame to a CSV file.
- `index=False` prevents the DataFrame index from being saved as an extra column.
- `pd.read_csv()` reads the CSV file back into a DataFrame.

### Expected Output

The student data will be displayed as a DataFrame.

### AI/ML Use

CSV files are commonly used to store datasets before preprocessing and model training.

### Data Analytics Use

CSV files are commonly used for storing and analyzing business data.

```python
import pandas as pd

data = {
    "Name": ["Arun", "Priya", "Kumar"],
    "Age": [22, 24, 23],
    "City": ["Chennai", "Madurai", "Coimbatore"]
}

df = pd.DataFrame(data)

# Write data to JSON
df.to_json("students.json", orient="records")

# Read data from JSON
df_read = pd.read_json("students.json")

print(df_read)
```

# 4. Parquet Files

## Step 1 – Understand the Concept

Parquet is a columnar file format designed for efficient storage and processing of large datasets.

It stores data by columns instead of storing complete rows together.

Pandas provides:

- `df.to_parquet()` – to write Parquet files
- `pd.read_parquet()` – to read Parquet files

### Why do we need Parquet?

- Efficient for large datasets.
- Provides good compression.
- Reads selected columns efficiently.
- Commonly used in data engineering and analytics.

## Step 2 – Demonstrate the Concept

### Simple Example

Create a DataFrame and save it as a Parquet file.

### Real-World Example

Large sensor datasets can be stored in Parquet format.

### Business Example

A company can store large sales datasets in Parquet for efficient analysis.

### AI/ML Use Case

Large machine learning datasets can be stored and loaded efficiently using Parquet.

### Data Analytics Use Case

Parquet is useful for analyzing large datasets because it is optimized for column-based operations.

```python
# Write data to Parquet
df.to_parquet("sales.parquet", index=False)

# Read data from Parquet
df_read = pd.read_parquet("sales.parquet")

print(df_read)
```

# 5. Pickle Files

## Step 1 – Understand the Concept

Pickle is a Python format used to serialize and save Python objects.

Serialization means converting a Python object into a format that can be stored and loaded later.

Pandas provides:

- `df.to_pickle()` – to save a DataFrame
- `pd.read_pickle()` – to load a DataFrame

### Why do we need Pickle?

- To save Python objects.
- To preserve Python data structures.
- To store processed DataFrames.
- To reuse data later without recreating it.

## Step 2 – Demonstrate the Concept

### Simple Example

Save a DataFrame as a Pickle file and load it again.

### Real-World Example

A Python application can save processed data for later use.

### Business Example

A company can save processed business datasets and reuse them in another Python program.

### AI/ML Use Case

Processed datasets and Python objects can be saved and loaded during ML workflows.

### Data Analytics Use Case

Analysts can save intermediate DataFrames and reuse them later.


## Step 3 – Implement the Concept (Python)

```python
import pandas as pd

data = {
    "Name": ["Arun", "Priya", "Kumar"],
    "Score": [85, 90, 78]
}

df = pd.DataFrame(data)

# Write data to Pickle
df.to_pickle("students.pkl")

# Read data from Pickle
df_read = pd.read_pickle("students.pkl")

print(df_read)
```

## Code Explanation

- `df.to_pickle()` saves the DataFrame as a Pickle file.
- `pd.read_pickle()` loads the saved Python object.
- The DataFrame structure is preserved.

### Expected Output

The original DataFrame will be displayed.

### AI/ML Use

Pickle can be used to store Python objects and processed data used in ML workflows.

### Data Analytics Use

Processed DataFrames can be saved and reused without repeating the preprocessing steps.

### Important Note

Only load Pickle files from trusted sources because Pickle files can execute arbitrary Python code when loaded.

# 6. SQL Tables

## Step 1 – Understand the Concept

SQL databases store data in tables.

Pandas can read SQL query results into a DataFrame using:

`pd.read_sql()`

Pandas can also write a DataFrame to a SQL table using:

`df.to_sql()`

### Why do we need SQL Table Operations?

- To work with database data.
- To analyze large structured datasets.
- To move data between databases and Pandas.
- To prepare database data for analysis and machine learning.

## Step 2 – Demonstrate the Concept

### Simple Example

Create a small SQL table and read it into Pandas.

### Real-World Example

A school database can store student information in SQL tables.

### Business Example

Companies store customer, sales, and transaction data in SQL databases.

### AI/ML Use Case

Machine learning datasets are often collected from production databases using SQL queries.

### Data Analytics Use Case

Analysts can query business databases and load the results into Pandas for analysis.


## Step 3 – Implement the Concept (Python)

```python
import pandas as pd
import sqlite3

# Create a temporary SQLite database
connection = sqlite3.connect("students.db")

data = pd.DataFrame({
    "Name": ["Arun", "Priya", "Kumar"],
    "Marks": [85, 90, 78]
})

# Write DataFrame to SQL table
data.to_sql("students", connection, if_exists="replace", index=False)

# Read SQL table into DataFrame
df = pd.read_sql("SELECT * FROM students", connection)

print(df)

connection.close()
```

## Code Explanation

- `sqlite3.connect()` creates a connection to a SQLite database.
- `data.to_sql()` writes the DataFrame into a SQL table.
- `pd.read_sql()` executes the SQL query and returns the result as a DataFrame.
- `connection.close()` closes the database connection.

### Expected Output

The student records stored in the SQL table will be displayed.

### AI/ML Use

ML pipelines can retrieve training data directly from SQL databases.

### Data Analytics Use

Analysts can query database tables and analyze the results using Pandas.

# 7. Reading Multiple Files

## Step 1 – Understand the Concept

Sometimes data is stored in multiple files instead of one file.

For example:

- January sales → `sales_jan.csv`
- February sales → `sales_feb.csv`
- March sales → `sales_mar.csv`

We can read multiple files and combine their data into one DataFrame.

Pandas provides `pd.concat()` to combine DataFrames.

### Why do we need this?

- To combine data from different files.
- To process monthly or yearly datasets.
- To create one complete dataset.
- To simplify analysis.

## Step 2 – Demonstrate the Concept

### Simple Example

Create multiple CSV files and combine them.

### Real-World Example

A school may store student data separately for each class.

### Business Example

A company may store monthly sales in separate CSV files.

### AI/ML Use Case

Training data may be distributed across multiple files and combined before preprocessing.

### Data Analytics Use Case

Multiple monthly or regional datasets can be combined for complete analysis.


## Step 3 – Implement the Concept (Python)

```python
import pandas as pd

# Create sample files
jan = pd.DataFrame({
    "Product": ["Laptop", "Phone"],
    "Sales": [60000, 30000]
})

feb = pd.DataFrame({
    "Product": ["Tablet", "Headphone"],
    "Sales": [20000, 5000]
})

jan.to_csv("sales_jan.csv", index=False)
feb.to_csv("sales_feb.csv", index=False)

# Read multiple files
df1 = pd.read_csv("sales_jan.csv")
df2 = pd.read_csv("sales_feb.csv")

# Combine DataFrames
combined_data = pd.concat([df1, df2], ignore_index=True)

print(combined_data)
```

## Code Explanation

- Two DataFrames are created and saved as separate CSV files.
- `pd.read_csv()` reads each file.
- `pd.concat()` combines the DataFrames row by row.
- `ignore_index=True` creates a new continuous index.

### Expected Output

The January and February sales records will appear in one DataFrame.

### AI/ML Use

Multiple data files can be combined to create a larger training dataset.

### Data Analytics Use

Monthly, regional, or departmental files can be combined for complete analysis.

# 8. File Path Handling

## Step 1 – Understand the Concept

A file path tells Python where a file is located.

There are two common types of paths:

- Relative path
- Absolute path

### Relative Path

A relative path starts from the current working directory.

Example:

`data/students.csv`

### Absolute Path

An absolute path gives the complete location of the file.

Example:

`C:/Users/User/Documents/data/students.csv`

### Why do we need File Path Handling?

- To locate files correctly.
- To organize datasets.
- To avoid FileNotFoundError.
- To make projects easier to manage.

## Step 2 – Demonstrate the Concept

### Simple Example

Use Python's `pathlib` to create and manage file paths.

### Real-World Example

A project may contain separate folders for raw data and processed data.

### Business Example

A company may organize files by year, month, department, or region.

### AI/ML Use Case

ML projects commonly have folders such as:

- `data/`
- `models/`
- `outputs/`

### Data Analytics Use Case

Data analysts can organize raw, cleaned, and final datasets into separate folders.


## Step 3 – Implement the Concept (Python)

```python
from pathlib import Path
import pandas as pd

# Create a data folder
data_folder = Path("data")
data_folder.mkdir(exist_ok=True)

# Create file path
file_path = data_folder / "students.csv"

# Create sample data
df = pd.DataFrame({
    "Name": ["Arun", "Priya", "Kumar"],
    "Marks": [85, 90, 78]
})

# Save the file
df.to_csv(file_path, index=False)

# Check whether the file exists
print("File exists:", file_path.exists())

# Read the file
df_read = pd.read_csv(file_path)

print(df_read)
```

## Code Explanation

- `Path()` creates a platform-independent file path.
- `mkdir()` creates the folder if it does not already exist.
- `/` is used to combine the folder path and file name.
- `exists()` checks whether the file exists.
- `pd.read_csv()` reads the file using the created path.

### Expected Output

The program will:

1. Create a `data` folder.
2. Create `students.csv` inside it.
3. Check that the file exists.
4. Read and display the student data.

### AI/ML Use

Proper file paths help organize datasets, models, and output files in ML projects.

### Data Analytics Use

File path handling helps analysts manage multiple datasets and project folders efficiently.
