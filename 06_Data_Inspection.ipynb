# 1. head()

## Step 1 – Understand the Concept

`head()` is used to view the first few rows of a Pandas DataFrame.

By default, `head()` displays the first 5 rows.

### Why is it needed?

When we load a large dataset, we do not need to display the entire dataset.
`head()` helps us quickly check whether the data has been loaded correctly and understand its structure.

## Step 2 – Demonstrate the Concept

### Simple Example
If a DataFrame contains 100 students, `head()` shows the first 5 students.

### Real-World Example
After loading customer data, we can use `head()` to quickly check the first few customer records.

### Business Example
A company can inspect the first few sales records to verify that product names, quantities, and sales values are loaded correctly.

### AI/ML Use Case
Before training a machine learning model, `head()` helps us understand the first few rows of the training dataset and identify the available features.

### Data Analytics Use Case
A data analyst can use `head()` to quickly inspect a dataset before performing analysis.


## Step 3: Implement the Concept

```python
import pandas as pd

df = pd.DataFrame({
    "Name": ["Arun", "Priya", "Kumar", "Anitha", "Rahul", "Divya"],
    "Age": [22, 24, 23, 25, 21, 26],
    "Marks": [85, 90, 78, 88, 76, 92]
})

print(df.head())
```

## Code Explanation

- The code creates a DataFrame containing student information.
- `head()` is used to display the first 5 rows.
- By default, `head()` returns 5 rows.
- We can also use `df.head(3)` to display the first 3 rows.

### AI/ML Use
It helps us quickly inspect the training or testing dataset before model development.

### Data Analytics Use
It helps analysts verify the first few records before performing data analysis.

# 11. value_counts()

## Step 1 – Understand the Concept

`value_counts()` counts how many times each unique value appears in a Series or column.

### Why is it needed?

It helps us understand the frequency of different values in a column.

## Step 2 – Demonstrate the Concept

### Simple Example
If a column contains:

`["Chennai", "Chennai", "Mumbai", "Delhi", "Chennai"]`

`value_counts()` shows that Chennai appears 3 times.

### Real-World Example
A company can count how many customers belong to each city.

### Business Example
A store can find how many times each product was purchased.

### AI/ML Use Case
It helps check the distribution of categories in a categorical feature and identify class imbalance.

### Data Analytics Use Case
Analysts use it to find the most common categories or values in a dataset.


## Step 3: Implement the Concept

```python
city_data = pd.DataFrame({
    "City": ["Chennai", "Chennai", "Mumbai", "Delhi", "Chennai", "Mumbai"],
    "Purchase": [500, 700, 300, 450, 800, 600]
})

print(city_data["City"].value_counts())
```

## Code Explanation

- The code creates a DataFrame containing city and purchase information.
- `city_data["City"]` selects the City column.
- `value_counts()` counts how many times each city appears.
- Chennai appears 3 times, Mumbai appears 2 times, and Delhi appears 1 time.

### AI/ML Use
It can be used to check category frequency and identify class imbalance in classification datasets.

### Data Analytics Use
It helps identify the most common categories and understand categorical data distribution.
